```python
class Atm: 
# constructor 
    def __init__(self):#Non parameterize constructor(not required parameter) 
        print(id(self))
        self.pin='123' 
        self.balance=10000
        # self.menu() 
    def menu(self): 
        user_input=input('''
         Hello, 
         How can I help you! 
         Please Choose Any one option.
         1.Press 1 to Balance Inquiry
         2.Press 2 to Case withdraw
         3.Press 3 to Pin Change
         4.Press 4 to Create Pin
         5.Any things else to exist''')
        if user_input=='1': 
            self.Balance_inquiry()
        elif user_input=='2':
            self.case_withdraw()
        elif user_input=='3': 
            self.Change_pin()
        elif user_input=='4': 
            self.create_pin() 
        else:
            exit()
    def create_pin(self): 
        user_pin=input("Enter Your Pin:")
        self.pin=user_pin
        user_balance=input("Enter Your Balance:")
        self.balance=user_balance
        print("pin created successfully")
        self.menu()
    def Change_pin(self): 
        old_pin=input("Enter your old pin:")
        
        if old_pin==self.pin: 
            new_pin=input("Enter new pin:")
            self.pin=new_pin
            print("New pin created successful")
            self.menu()
        else:
            print("Incorrect pin Please try again....")
            self.menu()
    def Balance_inquiry(self): 
        user_pin=input("Enter Your pin:")
        if user_pin==self.pin: 
            print("Your Balance is:",self.balance)
            self.menu()
        else: 
            print("Incorrect pin Please try again....")
            self.menu() 
    def case_withdraw(self): 
        user_pin=input("Enter Your pin:")
        if user_pin==self.pin: 
            req_bal=int(input("Enter Amount:"))
            if req_bal<=self.balance: 
                print("Collect Your Cash:")
                self.balance=self.balance-req_bal
                print("Remaing Balance is:",self.balance)
                self.menu()
            else: 
                print("Insufficient balance.....")
                self.menu()
        else:
            print("Incorrect pin please Try again.....")
            self.menu()
    
```

```python
class fraction: 
    def __init__(self,x,y): 
        self.num=x
        self.den=y
    # def __str__(self): 
    #     return '{}/{}'.format(self.num,self.den)
    def __add__(self,other): 
        new_num=self.num*other.den+other.num*self.den
        new_dem=self.den*other.den
        return '{}/{}'.format(new_num,new_dem)
    def __sub__(self,other): 
        new_num=self.num*other.den-other.num*self.den
        new_dem=self.den*other.den
        return '{}/{}'.format(new_num,new_dem) 
    def __mul__(self,other): 
        new_num=self.num*other.num
        new_dem=self.den*other.den
        return '{}/{}'.format(new_num,new_dem)
    def __truediv__(self,other): 
        new_num=self.num*other.den
        new_dem=self.den*other.num
        return '{}/{}'.format(new_num,new_dem)
    def convert_to_decimal(self):
        return self.num/self.den
        
    
        
```

```python
class Student: 
    def __init__(self,name,marks): 
        self.name=name
        self.__marks=marks
    def set_marks(self,updated_marks):
        if not isinstance(updated_marks,(int,float)): 
            raise TypeError('Marks must be Number.')
        if updated_marks <0 or updated_marks > 100: 
            raise ValueError('Marks must be between 0 to 100.')
        self.__marks=updated_marks
        print(f'Updated Marks is {self.__marks}')
    def get_marks(self): 
        return self.__marks
obj=Student('King',33)
obj.get_marks()
obj.set_marks(100)
```

```python
class Shape: 
    def area(self): 
        pass
        
class Circle(Shape): 
    def __init__(self,radius): 
        self.radius=radius
    def area(self):
        return 3.14 * self.radius**2
        
class Rectangle(Shape): 
    def __init__(self,length,width): 
        self.length=length
        self.width=width
    def area(self): 
        return self.length * self.width
class Square(Shape): 
    def __init__(self,side): 
        self.side=side
    def area(self): 
        return self.side**2 
shapes=[
    Circle(5), 
    Rectangle(2,4), 
    Square(5)
]
for shape in shapes: 
    print(shape.area())
```

```python
class Book:
    def __init__(self,title,author,is_issued=False): 
        self.title=title
        self.author=author
        self.is_issued=is_issued
class Library: 
    def __init__(self): 
        self.book=[]
    def add_book(self,book): 
        self.book.append(book)
        print(f'{book.title} added Sucessfully ')

    def issued_book(self,title): 
            for book in self.book: 
                if book.title==title: 
                    if book.is_issued: 
                        print('Already Issued')
                    else: 
                        book.is_issued=True
                        print('Book issued successfully')
                return
            print('Currently This Book is not Found')
    def return_book(self,title): 
        for book in self.book: 
            if book.title==title:
                if book.is_issued:
                    book.is_issued=False
                    print('Book was returned successfully')
                else:
                    print('Book was not issued from this library')
        print('Book not found')

        
b1 = Book("Python Basics", "John")
b2 = Book("Java Basics", "James")

lib = Library()

lib.add_book(b1)
lib.add_book(b2)

lib.issued_book("Python Basics")
lib.issued_book("Python Basics")
lib.return_book("Python Basics")

   
```

```python
from abc import ABC ,abstractmethod
class Vehicle(ABC): 
    @abstractmethod
    def calculate_rent(self,day):
        pass
        

class Car(Vehicle): 
    def calculate_rent(self,day): 
        return day*1000
class Bike(Vehicle): 
    def calculate_rent(self,day): 
        return day*300
# obj=Vehicle() # TypeError
obj1=Car()
obj2=Bike()
print(obj1.calculate_rent(3))
print(obj2.calculate_rent(3))
    
```

```python
class Cart: 
    def __init__(self,item=None): 
        self.item=item if item is not None else []
    def __add__(self,new_item): 
        return Cart(self.item+new_item.item)
    def __len__(self): 
        return len(self.item)
cart1=Cart(['pen'])
cart2=Cart(['book'])
cart3=cart1+cart2
print(cart3.item) 
print(len(cart3))
```

```python
class Database:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
db1 = Database()
db2 = Database()
db3 = Database()

print(db1 is db2)  
print(db2 is db3)  
```
