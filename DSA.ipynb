```python
class stack:
    def __init__(self):
        self.stack=[]
    def push(self,element): 
        if self.stack is not None: 
            self.stack.append(element) 
        else: 
            print("welecome") 
    def pop(self): 
        if len(self.stack)>0: 
            return self.stack.pop() 
        else: 
            #raise IndexError("stack is empty") 
            #raise Exception("stack is empty")
            print("stack is empty") 
obj = stack() 
print(obj.stack) 
print(obj.pop())
```
