```python
class stack:
    def __init__(self):
        self.stack=[]
    def push(self,element): 
        if self.stack is not None: 
            self.stack.append(element) 
        else: 
            print("welecome") 
    def pop(self): 
        if len(self.stack)>0: 
            return self.stack.pop() 
        else: 
            #raise IndexError("stack is empty") 
            #raise Exception("stack is empty")
            print("stack is empty") 
obj = stack() 
print(obj.stack) 
print(obj.pop())
```

## Time and Space complexity 

```python
# Q1: print_sum(n) recursion se 1 se n tak sum nikalo.

def add(n):
    if n==0: 
        return 0
    return n + (add (n-1) )
print(add(5))

# Time Complexity: O(n)
# Space Complexity: O(n), due to recursion call stack.
```

```python
# Q2: List me sabse bada element dhoondo, Complexity kya hogi? O(1) ya O(n)?
l=[1,6,8,3,5,8,10,4,74,14,57,97,34,6,0]
maximum=l[0]
for i in l: 
    if i > maximum: 
        maximum=i 
print(maximum)
# Time Complexity: O(n)
# Kyunki worst case mein hume list ke saare elements check karne padte hain.
```

```python
# Q3: [1,2,2,3,3,3] me Counter se sabse zyada baar aane wala element kaunsa hai?
from collections import Counter
l=[1,2,2,3,3,3]
count=Counter(l)
print(count)
print(count.most_common(1))
print(count.most_common(1)[0][0])
```

```python
# Q4: Bina max() function use kiye list ka second largest element find karo.
l=[2,65,3,6,4,5,8,3,22,5,87]
maximum=l[0]
sec_max=0
for i in l: 
    if i > maximum:
        sec_max=maximum
        maximum=i
    elif i > sec_max and i != maximum: 
        sec_max=i
print(f'Largest value is: {maximum}')
print(f'Second Largest value is: {sec_max}')

        
```

```python
def is_palindrome(arr): 
    j = len(arr)-1
    i = 0
    while j > i : 
        if arr[i] != arr[j]:
            return False
        i+=1
        j-=1
    return True
is_palindrome([1,2,3,2,1,3])    
```

```python
n = 20
primes = []
for num in range(2, n + 1):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(num)
print(primes)
```
