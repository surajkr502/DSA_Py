# Recursion:Fn Calling it self
Types of Recursion: 
1. Direct Recursion
2. Indirect Recursion
3. Head Recursion
4. Tail Recursion
5. Recusive Tree

1. Base Condition
2. Recusion Call or Recursion relationship

```python
def loop(n): #tail recursion
    if n == 0: #Base condition 
        return
    print(n)
    loop(n-1) #recursion call
loop(5)
```

```python
def loop2(n): # Head recursion 
    if n==0: 
        return
    loop2(n-1)
    print(n)
loop2(5)
    
```

```python
def table(n,s): 
    if s > 10: 
        return
    print(n*s) 
    table(n,s+1)
table(4,1)
```

```python
fact = 1
for i in range(1,6):
    fact= i*fact
print(fact)
    
    
```

```python
def factorial(n):
    if n == 0: 
        return 1
    return n * factorial(n-1)
factorial(5)
```

```python
def power(base,p): 
    if p == 0: 
        return 1
    power(base,p-1)
    print(base**p)
power(5,4)
```

```python
n = 4
for i in range(1,n): 
    print(n**i)
```

```python
# Indirect call
def loop2(n): # Head recursion 
    if n==0: 
        return
    loop2(n-1)
    print(n)
loop2(5)
    
```

```python
# Indirect call
def recall(p): 
    loop2(p) #indirectly calling loop2 recursion 
    print('completed')
recall(5)
```

```python
# skew tree 
def pw(b,s): 
    if s == 0: 
        return 1
    return b * pw(b,s-1)
pw(2,3)
```

```python
# fibonacci series
p = 0 
c = 1
print(p,c,end=" ")
for i in range(1,7): 
    n = p+c 
    print(n,end=' ')
    p = c 
    c = n 
    
```

```python
def fn(n): 
    if n == 0 or n == 1: 
        return n 
    return fn(n-1) + fn(n)
```
