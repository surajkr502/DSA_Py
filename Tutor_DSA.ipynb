## Extraction of digits


```python
# Extract the digits

n=int(input('Enter integer:'))
num=n 
while num > 0: 
    last_digit=num%10
    print(last_digit)
    num=num//10
```

```python
# Count the digits 
try: 
    n = int(input('Enter Integer Number: '))
    print(f'You entered a valid integer: {n}')
    temp=abs(n)
    if temp==0: 
        count=1
    else: 
        count=0
        while temp>0: 
            temp//=10
            count+=1 
    print(f'Total count:{count}')
except ValueError: 
    print('Error: That was not a valid integer! Please enter numbers only.')

    
```

```python
# palindrome number
n=121
num=n
result=0
while num>0: 
    rem=num%10
    result=(result*10)+rem
    num//=10
if result == n: 
    print(f'Number {n} is palindrome.')
else: 
    print(f'Number {n} is not a palindrome')
    
    
    
```

```python
def is_palindrome(value):
    value = str(value)

    if value == value[::-1]:
        print(f"{value} is a palindrome.")
    else:
        print(f"{value} is not a palindrome.")

is_palindrome(121)
is_palindrome("madam")
is_palindrome("hello")
```

```python
# Armstronge Number
n=153
num = n
str_n = str(n)
len_n=len(str_n)
result=0
while num > 0: 
    digit=num%10
    result+= digit**len_n
    num//=10
if n==result: 
    print(f'Given Number {n} is Armstronge number')
else: 
    print(f'Given Number {n} is not an Armstronge number')
   
```

```python
# Factor/Divisor 
n=10
factor=[]
for i in range(1,n+1): 
    if n%i == 0: 
        factor.append(i)
print(factor)

# Time Complexity= 0(n)
```

```python
n = 10
factors = []

for i in range(1, int(n**0.5) + 1):
    if n % i == 0:
        factors.append(i)

        if i != n // i:
            factors.append(n // i)

print(sorted(factors))

# Optimized Time Complexity: O(√n)
# Space Complexity: O(√n)
```
