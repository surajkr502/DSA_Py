## Extraction of digits


```python
# Extract the digits

n=int(input('Enter integer:'))
num=n 
while num > 0: 
    last_digit=num%10
    print(last_digit)
    num=num//10
```

```python
# Count the digits 
try: 
    n = int(input('Enter Integer Number: '))
    print(f'You entered a valid integer: {n}')
    temp=abs(n)
    if temp==0: 
        count=1
    else: 
        count=0
        while temp>0: 
            temp//=10
            count+=1 
    print(f'Total count:{count}')
except ValueError: 
    print('Error: That was not a valid integer! Please enter numbers only.')

    
```

```python
# palindrome number
n=121
num=n
result=0
while num>0: 
    rem=num%10
    result=(result*10)+rem
    num//=10
if result == n: 
    print(f'Number {n} is palindrome.')
else: 
    print(f'Number {n} is not a palindrome')
    
    
    
```

```python
def is_palindrome(value):
    value = str(value)

    if value == value[::-1]:
        print(f"{value} is a palindrome.")
    else:
        print(f"{value} is not a palindrome.")

is_palindrome(121)
is_palindrome("madam")
is_palindrome("hello")
```

```python
# Armstronge Number
n=153
num = n
str_n = str(n)
len_n=len(str_n)
result=0
while num > 0: 
    digit=num%10
    result+= digit**len_n
    num//=10
if n==result: 
    print(f'Given Number {n} is Armstronge number')
else: 
    print(f'Given Number {n} is not an Armstronge number')
   
```

```python
# Factor/Divisor 
n=10
factor=[]
for i in range(1,n+1): 
    if n%i == 0: 
        factor.append(i)
print(factor)

# Time Complexity= 0(n)
```

```python
n = 10
factors = []

for i in range(1, int(n**0.5) + 1):
    if n % i == 0:
        factors.append(i)

        if i != n // i:
            factors.append(n // i)

print(sorted(factors))

# Optimized Time Complexity: O(√n)
# Space Complexity: O(√n)
```

## Hashing in python

```python
# using for loop 
s=[4,2,6,9,6,9,5,2,4,5,8]
m=[5,6,4,3,6,7,54,21]
hash_list=[0]*11
for num in s: 
    hash_list[num]+=1 
for num1 in m: 
    if num1<1 or num1 >10: 
        print(0)
    else: 
        print(hash_list[num1])
# print(hash_list)
# time complexity: O(n*n)
```

```python
# Number count using dict_hashing 
l1=[2,4,5,6,3,2]
l2=[5,6,3,4,66,89]
hash_map=dict()
for num in l1: 
    hash_map[num]=hash_map.get(num,0)+1
for num in l2: 
    print(num, ':' ,hash_map.get(num,0))

```

```python
# Brute force Solution
s1="yduhdehdhdjsh"
s2=['d','y','h','a']
hash_map={}
for s in s1: 
    hash_map[s]=hash_map.get(s,0)+1
for s in s2: 
    print(s, ":" , hash_map.get(s,0))
    
```

```python
# Optimize solution 
# assume all characters are in small letter

s1="yduhdehdhdjshza"
s2=['d','y','h','a']
hash_list=[0]*26
for ch in s1: 
    ascii_val=ord(ch)
    index=ascii_val-97
    hash_list[index]+=1
for ch in s2: 
    ascii_val=ord(ch)
    index=ascii_val-97
    print(hash_list[index])
    
```

```python
# Handling all characters
s1 = "yduhdehdhdjshza$^*%@!$#"
s2 = ['d', 'y', 'h', 'a', '&', '@']
hash_list = [0] * 128
for ch in s1:
    ascii_val = ord(ch)
    index = ascii_val
    hash_list[index] += 1

for ch in s2:
    ascii_val = ord(ch)
    index = ascii_val
    print(ch, ":", hash_list[index])
```

## Recursion 
Recursion mein do important things hoti hain:
1. Base Case: Recursion ko stop karne ki condition.
2. Recursive Case: Function khud ko call karta hai.

```python
# Head Recursion
count=0
def greet():
    global count
    if count==4: 
        return
    print('Hello!')
    count+=1
    greet()

greet()
```

```python
def greet(count): 
    if count==4: 
        return
    print('Hello!')
    greet(count+1)
greet(0)
```

```python
# Tail Recursion
def greet(count): 
    if count==4: 
        return
    greet(count+1)
    print('Hello!')
greet(0)

# TC: O(n+1) 
# SC: O(n+1)
```

```python
def fun(x,n): 
    if n==0: 
        return
    print(x)
    fun(x,n-1)
fun(15,5)
```

```python
def fun(i,n): 
    if n==0: 
        return
    print(i)
    fun(i+1,n-1)
fun(1,5)
```

```python
# Paramiterized recursion 
def fun(sum,i,n): 
    if i > n: 
        print(sum)
        return
    fun(sum+i,i+1,n) 
fun(0,1,5)
```

```python
# Function recursion 
def fun(n): 
    if n == 1: 
        return 1 
    return n + fun(n-1)
print(fun(5))
```

```python
def fun(n): 
    if n == 1 or n == 0: 
        return 1
    return n * fun(n-1)
print(fun(0))


# TC:- O(n) 
# SC:- O(n) stack space
```
