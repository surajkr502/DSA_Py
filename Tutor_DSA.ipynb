```python
def is_palindrome(value):
    value = str(value)

    if value == value[::-1]:
        print(f"{value} is a palindrome.")
    else:
        print(f"{value} is not a palindrome.")

is_palindrome(121)
is_palindrome("madam")
is_palindrome("hello")
```

```python
n = 10
factors = []

for i in range(1, int(n**0.5) + 1):
    if n % i == 0:
        factors.append(i)

        if i != n // i:
            factors.append(n // i)

print(sorted(factors))

# Optimized Time Complexity: O(√n)
# Space Complexity: O(√n)
```

## BubbleSorting

```python
class Solution:
    def bubbleSort(self, arr):
        n = len(arr)
        for i in range(n-1, 0, -1):
            is_sorted = True  
            for j in range(i):
                if arr[j] > arr[j+1]:
                    arr[j], arr[j+1] = arr[j+1], arr[j]
                    is_sorted = False  
            if is_sorted == True:
                break    
        return arr
obj = Solution()
print(obj.bubbleSort([3,2,4,2,7,8,0,7,8,4,2,4]))
```

```python
arr=[5,3,9,1,4,8,2,0,7,6]
n=len(arr)
for i in range(1,n): 
    key=arr[i]
    j=i-1
    while j >= 0 and arr[j] > key: 
        arr[j+1]=arr[j]
        j-=1
    arr[j+1]=key
print(arr)

# TC:- O(N**2)
# SC:- O(1)
```

```python
class Solution:
    def insertionSort(self, arr):
        n=len(arr)
        for i in range(1,n): 
            key=arr[i]
            j=i-1
            while j >= 0 and arr[j] > key: 
                arr[j+1] = arr[j]
                j-=1
            arr[j+1] = key
        return arr
obj=Solution()
print(obj.insertionSort([6,4,9,1,0,2,7,3,8]))
```

```python

class Solution:
    def mergeSort(self, arr, l, r):

        if l >= r:
            return

        mid = (l + r) // 2

        left_arr = arr[l:mid + 1]
        right_arr = arr[mid + 1:r + 1]

        self.mergeSort(left_arr, 0, len(left_arr) - 1)
        self.mergeSort(right_arr, 0, len(right_arr) - 1)

        merged = merge_arr(left_arr, right_arr)

        arr[l:r + 1] = merged


def merge_arr(left, right):
    result = []
    i, j = 0, 0
    n, m = len(left), len(right)

    while i < n and j < m:
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    while i < n:
        result.append(left[i])
        i += 1

    while j < m:
        result.append(right[j])
        j += 1

    return result
arr = [4, 1, 3, 9, 7]
obj = Solution()
obj.mergeSort(arr, 0, len(arr) - 1)
print(arr)
# Time: O(n log n)
# Auxiliary space: O(n) for temporary arrays and merge storage.
```

## Quick Sort

```python
def partitons(arr,low,high): 
    pivot=arr[low]
    i,j=low,high
    while i < j: 
        while arr[i] <= pivot and i <= high - 1: 
            i+=1
        while arr[j] > pivot and j >= low + 1:
            j+=1
        if i < j:
            arr[i],arr[j] = arr[j],arr[i] 
    arr[low],arr[j] = arr[j],arr[low]
    return i 
    
    
    
```

```python
def quick_sort(arr,low,high): 
    if low < high: 
        p_index=partitons(arr,low,high)
        quick_sort(arr,low,p_index-1)
        quick_sort(arr,p_index+1,high)
```

```python
class Solution:
    def quickSort(self, arr, low, high):
        if low < high:
            p_index = self.partition(arr, low, high)

            self.quickSort(arr, low, p_index - 1)
            self.quickSort(arr, p_index + 1, high)

    def partition(self, arr, low, high):
        pivot = arr[low]

        i, j = low, high

        while i < j:

            while i <= high - 1 and arr[i] <= pivot:
                i += 1

            while j >= low + 1 and arr[j] > pivot:
                j -= 1

            if i < j:
                arr[i], arr[j] = arr[j], arr[i]

        arr[low], arr[j] = arr[j], arr[low]

        return j


arr = [4, 1, 3, 9, 7]

obj = Solution()
obj.quickSort(arr, 0, len(arr) - 1)

print(arr)

# 1. Best case	O(n log n)
# 2. Average case	O(n log n)
# 3. Worst case	O(n²)
```
